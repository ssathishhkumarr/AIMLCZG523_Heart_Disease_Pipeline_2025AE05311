```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style for professional charts
sns.set_theme(style="whitegrid")

# Load the cleaned dataset
df = pd.read_csv("../data/heart_disease.csv")
df.head()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(x="target", hue="target", data=df, palette="viridis", legend=False)
plt.title("Heart Disease Class Balance (0 = Absence, 1 = Presence)")
plt.xlabel("Target")
plt.ylabel("Count")
plt.savefig("../screenshots/class_balance.png", bbox_inches='tight') # Saves for your report
plt.show()
```

```python
numeric_features = ["age", "trestbps", "chol", "thalach", "oldpeak"]
df[numeric_features].hist(figsize=(12, 8), bins=20, color='teal', edgecolor='black')
plt.suptitle("Histograms of Key Numerical Features", fontsize=16)
plt.savefig("../screenshots/histograms.png", bbox_inches='tight') # Saves for your report
plt.show()
```

```python
plt.figure(figsize=(12, 10))
correlation_matrix = df.corr()
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Feature Correlation Heatmap", fontsize=16)
plt.savefig("../screenshots/correlation_heatmap.png", bbox_inches='tight') # Saves for your report
plt.show()
```
